```python
import pandas as pd
```

```python
# Messy Data
data = {
    'Name': ['Palak', 'Riya', ' palak ', 'Aman', 'Riya', None, 'AMAN'],
    'Age': [22, None, 22, 25, 24, 23, 25],
    'City': ['Bhopal', 'Indore', 'bhopal', 'Delhi', 'Indore', 'Delhi', 'DELHI'],
    'Salary': [30000, 25000, 30000, None, 25000, 40000, 50000]
}
df = pd.DataFrame(data)
df
```

```python

# Cleaning Automation - FIXED CODE
df['Name'] = df['Name'].fillna('Unknown')
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

df.drop_duplicates(inplace=True)

df['Name'] = df['Name'].astype(str).str.strip().str.title()
df['City'] = df['City'].astype(str).str.strip().str.title()

df
```

```python
# Automated Reporting
print("--- AUTOMATED CLEANING REPORT ---")
print(f"Total Records after cleaning: {len(df)}")
print(f"Missing Names: Fixed with 'Unknown'")
print(f"Missing Age/Salary: Fixed with Mean value")
print(f"Duplicates: Removed")
print(f"Inconsistent City/Name: Fixed (bhopal/DELHI -> Bhopal/Delhi)")
print("\nFinal Summary:")
print(df.describe())
```

```python
import matplotlib.pyplot as plt

# Visual Summary - Automated
plt.figure(figsize=(6,4))
df['City'].value_counts().plot(kind='bar', color=['skyblue','orange','green'])
plt.title('Employees by City - Cleaned Data Report')
plt.xlabel('City')
plt.ylabel('Count')
plt.show()

plt.figure(figsize=(6,4))
plt.hist(df['Salary'], bins=5, color='green')
plt.title('Salary Distribution')
plt.show()
```
